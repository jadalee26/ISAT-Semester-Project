# Background

### Hurricane Maria Overview

Hurricane Maria was one of the most destructive Atlantic hurricanes in modern Caribbean history, making landfall in Puerto Rico on September 20, 2017, as a high-end Category 4 storm with sustained winds of approximately 155 mph. The storm rapidly intensified before landfall and produced widespread devastation across Puerto Rico and neighboring Caribbean islands. Maria became one of the strongest hurricanes to strike Puerto Rico in nearly a century and caused catastrophic damage throughout the island.

### Physical Drivers of Damage

The extreme destruction caused by Hurricane Maria resulted from a combination of intense wind speeds, heavy rainfall, storm surge, and prolonged exposure to severe weather conditions. Wind damage destroyed buildings, uprooted vegetation, and severely damaged power and communication infrastructure. Heavy rainfall triggered widespread flooding and landslides across mountainous regions of Puerto Rico, while storm surge inundated coastal communities and contributed to erosion and property destruction. These combined meteorological impacts created both immediate and cascading environmental hazards.

### Human and Environmental Consequences

Hurricane Maria caused severe humanitarian and environmental consequences across Puerto Rico. The storm collapsed the island’s electrical grid, leaving nearly all residents without power for extended periods, in some areas for several months. Water treatment and distribution systems were also heavily damaged, limiting access to clean drinking water. Transportation infrastructure including roads, bridges, and airports suffered major disruptions, slowing emergency response and recovery efforts.

The hurricane contributed to an estimated 2,975 deaths in Puerto Rico, many resulting indirectly from prolonged infrastructure failure, limited medical access, contaminated water supplies, and delayed emergency services. Environmentally, the storm caused landslides, coastal erosion, deforestation, habitat destruction, and contamination of freshwater systems through runoff and infrastructure damage.

### Vulnerability and Broader Significance

Puerto Rico’s vulnerability to Hurricane Maria was amplified by preexisting economic and infrastructure challenges. Prior to the storm, much of the island’s electrical grid, water systems, and transportation infrastructure suffered from underinvestment and deferred maintenance. These systemic weaknesses increased the severity and duration of the disaster’s impacts.

More broadly, Hurricane Maria highlights the increasing risks posed by extreme tropical cyclones in a changing climate. Warmer sea surface temperatures and rising sea levels are expected to increase the intensity and destructiveness of hurricanes, particularly for coastal and island communities. As a result, Hurricane Maria serves as a critical example of the intersection between environmental hazards, infrastructure vulnerability, and climate resilience.

### Research Need

Given the widespread destruction and long-term impacts caused by Hurricane Maria, quantitative analysis of storm intensity, infrastructure damage, mortality, and recovery metrics is essential for understanding disaster vulnerability and improving future resilience planning. This project investigates Hurricane Maria as a case study to evaluate the environmental and societal impacts of extreme weather events and to identify patterns that can inform future disaster preparedness and infrastructure planning.

### Data Analysis and Environmental Impact
The analysis of Hurricane Maria reveals a catastrophic synergy between atmospheric mechanics and record-breaking environmental drivers. Central to this volatility is a notable "Thermal Rise" within the Caribbean, where ocean temperature anomalies reached a record +1.3°C in 2024. This elevated "energy floor" provided the high ocean heat necessary to fuel Maria’s "Rapid Growth," a phenomenon documented in the storm’s tracking data. The best-track metrics reveal a staggering 110 kt intensity jump in only 48 hours, as the storm escalated from a 40 kt tropical storm on September 16 to a high-end Category 5 powerhouse with 150 kt winds and a minimum pressure of 908 mb by September 20.
By synthesizing these tracking milestones with multidimensional precipitation NetCDF files, the research identifies a critical "Hazard Sync." During the storm's landfall near Yabucoa, Puerto Rico, peak rainfall intensity (0.007 mm) aligned perfectly with maximum sustained winds, creating a simultaneous flood and wind threat that overwhelmed local infrastructure. The datasets, processed using xarray and Matplotlib, further illustrate the "persistence" of these effects; sustained ocean heat allowed for continued heavy rainfall even after wind speeds began to decline post-landfall.
These findings support a sobering "New Standard" for the region. Long-term trends analyzed by NASA and NCEI suggest that the extreme severity witnessed in 2017 is transitioning from an outlier to a regional baseline. Ultimately, while the research navigated challenges in data availability, integrating these metrics according to FAIR Principles underscores a definitive conclusion: rising global surface temperatures have fundamentally shifted the scale of hurricane destruction, making open-science data and climate resilience more urgent than ever.
